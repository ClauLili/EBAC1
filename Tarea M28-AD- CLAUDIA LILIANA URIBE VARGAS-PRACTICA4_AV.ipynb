LEER EL ARCHIVO Y GUARDARLO EN UNA LISTA Amazon_top100_bestselling_books_2009to2021.xlsx

```python
# lectura de archivo xlsx
import openpyxl
from pathlib import Path

xlsx_file = Path('.', 'Amazon_top100_bestselling_books_2009to2021.xlsx')
wb_obj = openpyxl.load_workbook(xlsx_file)

# Read the active sheet:
sheet = wb_obj.active
```

```python
# para obtener las 10 primeras lineas del archivo
for row in sheet.iter_rows(min_row=1,
                           max_row=10,
                           values_only=True):
    print(row)
```

```python
# mostrar las ultimas 10 lineas
total_filas = sheet.max_row
for row in sheet.iter_rows(min_row=total_filas-9,
                           max_row=total_filas,
                           values_only=True):
    print(row)
```

```python
# obtener las 4 primeras columnas y generar un nuevo archivo
for row in sheet.iter_rows(min_row=1,
                           max_col=4,
                           values_only=True):
    print(row)

nuevo_libro = openpyxl.Workbook()

nueva_hoja = nuevo_libro.active

for row in sheet.iter_rows(min_row=1,
                           max_col=4,
                           values_only=True):

    nueva_hoja.append(row)
    nuevo_libro.save("nuevo_top100_bestselling_books.xlsx")
```

LEER EL ARCHIVO Y GENERAR UN ARCHIVO NUEVO GOT Subtitles.json

```python
import json
from pathlib import Path
```

```python
# se lee la carpeta que contiene 7 archivos json 
carpeta_json = Path(r"C:\Users\Claudia\Downloads\GOT Subtitles")

archivos_json = list(carpeta_json.glob("*.json"))

print(archivos_json)
```

```python
# se juntan la informacion en un solo archivo
datos_got = []

for archivo in archivos_json:
    with open(archivo, "r", encoding="utf-8") as file:
        datos = json.load(file)
        datos_got.append(datos)

print("Archivos JSON leídos:", len(datos_got))
```

```python
# Se usa dumps de la librería json para pasar la salida a formato json y leer su estructura
print(json.dumps(datos_got[0], indent=4, ensure_ascii=False))
```

```python
# se muestran los 10 primeros
print(json.dumps(datos_got[:10], indent=4, ensure_ascii=False))
```

```python
# se muestran los 10 ultimos
print(json.dumps(datos_got[-10:], indent=4, ensure_ascii=False))
```

```python
# sirve para crear un nuevo archivo y guardar todos los archivos en uno solo
with open("GOT_Subtitles_Completo.json", "w", encoding="utf-8") as file:
    json.dump(datos_got, file, indent=4, ensure_ascii=False)
```
