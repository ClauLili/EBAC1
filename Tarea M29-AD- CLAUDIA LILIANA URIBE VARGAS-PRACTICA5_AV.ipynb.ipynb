```python
#importar librerias

import requests
from bs4 import BeautifulSoup
import json
```

```python
#Importación de sitio web con requests e impresión del contenido (cualquier sitio) 

url = "https://www.uaq.mx"

respuesta = requests.get(url)

print("Código de respuesta:", respuesta.status_code)
print(respuesta.text[:1000])
```

```python
#Uso de librería BeautifulSoup y requests
soup = BeautifulSoup(respuesta.text, "html.parser")

print(soup.prettify()[:1000])
```

```python
##  Extracción de textos del sitio web
titulos = soup.find_all(["h1", "h2", "h3"])

for titulo in titulos:
    texto = titulo.get_text(strip=True)
    if texto:
        print(texto)
```

```python
# Se hace Scraping de una página de libros Se extraen libros, precios y disponibilidad.
url_libros = "https://books.toscrape.com/"

respuesta_libros = requests.get(url_libros)

print("Código de respuesta:", respuesta_libros.status_code)

soup_libros = BeautifulSoup(respuesta_libros.text, "html.parser")

libros = soup_libros.find_all("article", class_="product_pod")

for libro in libros:
    titulo = libro.h3.a["title"]
    precio = libro.find("p", class_="price_color").get_text(strip=True)
    disponibilidad = libro.find("p", class_="instock availability").get_text(strip=True)

    print("Título:", titulo)
    print("Precio:", precio)
    print("Disponibilidad:", disponibilidad)
    print("-" * 40)
```

```python
# Generación de un listado simple, ordenado, de los ítems que aparecen en las listas
import requests
from bs4 import BeautifulSoup

# URL de prueba para scraping
url = "https://books.toscrape.com/"

# Obtener contenido
respuesta = requests.get(url)

# Parsear HTML
soup = BeautifulSoup(respuesta.text, "html.parser")

# Buscar libros
libros = soup.find_all("article", class_="product_pod")

# Crear lista vacía
titulos = []

# Extraer títulos
for libro in libros:
    titulo = libro.h3.a["title"]
    titulos.append(titulo)

# Ordenar lista
titulos_ordenados = sorted(titulos)

# Mostrar resultados
for i, titulo in enumerate(titulos_ordenados, start=1):
    print(f"{i}. {titulo}")
```

```python
import requests
import json

# API pública de prueba
url_api = "	https://the-rosary-api.vercel.app/v1/today"

# Realizar petición
respuesta_api = requests.get(url_api)

# Validar respuesta
print("Código de respuesta:", respuesta_api.status_code)

# Convertir respuesta a JSON
datos = respuesta_api.json()

# Mostrar JSON ordenado
print(json.dumps(datos, indent=4))
```

```python
# Extraer información específica de la API

for usuario in datos:
    print("ID:", usuario["id"])
    print("Fecha_rosario:", usuario["rosary_date"])
    print("Dia_rosario:", usuario["rosary_day"])
    print("-" * 40)
```
