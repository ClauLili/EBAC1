```python
# Convertir 2 columnas  a numericas
df['Release Clause'] = df['Release Clause'].replace({'€':'', 'M':'', 'K':''}, regex=True)
df['Release Clause'] = pd.to_numeric(df['Release Clause'], errors='coerce')
df['Wage'] = df['Wage'].replace({'€':'', 'K':''}, regex=True)
df['Wage'] = pd.to_numeric(df['Wage'], errors='coerce')
```

```python
df.describe()
```

```python
# Añadir columna Years Playing

df['Joined'] = pd.to_datetime(df['Joined'], errors='coerce')
df['Years Playing'] = 2026 - df['Joined'].dt.year
```

```python
# Ordenar datos por Release Clause
df.sort_values(by='Release Clause', ascending=False)
```

```python
# ❖Opcional❖: Generar un gráfico que contenga, por año, el número de jugadores
df_temp = df.groupby(df['Joined'].dt.year).size().rename('players').reset_index()
df_temp = df_temp.rename(columns={'Joined': 'joined_year'})

df_temp
```
